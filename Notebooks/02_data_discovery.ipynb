## Comparaison du nombre de lignes avant et après suppression des doublons

```python
df_sans_doublons = df.drop_duplicates()

print("Nombre de lignes avant :",len(df))
print("Nombre de lignes après :",len(df_sans_doublons))
```

## Calcul du pourcentage de valeurs manquantes par colonne

```python
def pourcentage_VM(df):
    return (df.isnull().sum() / len(df)) * 100

pourcentage_VM(df)
```
