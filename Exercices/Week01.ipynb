## 1. la moyenne d'une liste

```python
locations_velos=[120,150,180,200,250]
sum(locations_velos)
len(locations_velos)
print("la moyenne du liste est :",sum(locations_velos)/len(locations_velos))
```

## 2. comptage des valeurs sup à un seuil

```python
locations_velos=[120,150,180,200,250] 
seuil= int(input ("Entrer un seuil:"))
cpt=0
for i in range(len(locations_velos)):
    if locations_velos[i]>seuil :
        cpt=cpt+1
print("Nombrer de valeurs supérieur à",seuil, ":",cpt)
```

## 3.Description d'une heure de la journée 

```python
def describe_heure(heure):
    if 5<= heure <=11:
        return "Matin"
    elif 12<= heure <=17:
        return "Après-midi"
    elif 18<= heure <=21:
        return "soir"
    else:
        return "Nuit" 
## controle de saisie 
while True:
   heure = int (input("Entrer une heure (0-23):"))
   if 0<= heure <=23:
    break
   print("heure invalide ! Réssayez.") 

print(describe_heure(heure))
```

## 4.une fonction retournant la moyenne d’une liste vide ou non vide

```python
def moyenne(liste):
    if len(liste) == 0:
        return 0

    moyenne = sum(liste) / len(liste)
    return moyenne
locations_velos = [120, 150, 180, 200, 250]

print("La moyenne est :",moyenne(locations_velos))
```

## 5.Vérification des heures (7h–9h et 17h–19h)

```python
while True:
    heure = int(input("Entrer une heure entre 0 et 23 : "))

    if 0 <= heure <= 23:
        break

    print("Heure invalide ! Réessayez.")

if 7 <= heure <= 9 or 17 <= heure <= 19:
    print("L'heure appartient à l'un des deux intervalles.")
else:
    print("L'heure n'appartient à aucun des deux intervalles.")
```
